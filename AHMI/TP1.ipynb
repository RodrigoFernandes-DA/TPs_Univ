```python
class automaton : 
    def __init__(self, lang):
        self.lang = lang
        self.ini = idle
        self.sets = []
        for i in self.lang:
            # if i in ini.trans:
                if i == "coin" : 
                    out = coin.output
                    ini = coin.inpu
                    self.sets.append(ini)
                    print(f"from {ini.id} to {out.id}")
                elif i == "push" : 
                    out = push.output
                    ini = push.inpu
                    self.sets.append(ini)
                    print(f"from {ini.id} to {out.id}")
                elif i == "coffee" : 
                    out = coffee.output
                    ini = coffee.inpu
                    self.sets.append(ini)
                    print(f"from {ini.id} to {out.id}")
            #else : print("i impossible")
        print(self.sets)
    
class state : 
    def __init__(self, id, trans):
        self.id = id
        self.trans = trans
    
class transition : 
    def __init__(self, label, inpu, output):
        self.label = label
        self.inpu = inpu
        self.output = output
```

```python
idle = state("idle", ["coin"])
payed = state("payed", ["back", "push"])
returning = state("back", ["money"])
serving = state("serving", ["coffee"])

coin = transition("coin", idle, payed)
back = transition("back", payed, returning)
money = transition("money", returning, idle)
push = transition("push", payed, serving)
coffee = transition("coffee", serving, idle)
```
