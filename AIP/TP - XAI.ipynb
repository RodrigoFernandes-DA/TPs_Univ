```python
def load_imagenet_labels():
    """Load ImageNet class labels"""
    url = "https://raw.githubusercontent.com/anishathalye/imagenet-simple-labels/master/imagenet-simple-labels.json"
    response = requests.get(url)
    return eval(response.text)

imagenet_labels = load_imagenet_labels()

print(imagenet_labels[:10])
```

## a) Applying at least 3 attribution methods on a sample of images

```python
# Pour afficher une heatmap d'attribution
def normalize_attr(attr):
    # attr: (H,W) ou (H,W,3)
    a = attr.copy()
    a = a - a.min()
    if a.max() > 0:
        a = a / a.max()
    return a

def show_attributions(pil_img, attr_map, title="Attribution"):
    # attr_map en (H,W)
    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1)
    plt.imshow(pil_img)
    plt.title("Image")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(attr_map, cmap="inferno")
    plt.title(title)
    plt.axis("off")
    plt.show()

# 3 méthodes minimum : Occlusion, IntegratedGradients, InputXGradient
def compute_attributions(model, x, target,
                         occ_patch=(15, 15),  # taille du patch masqué (H,W)
                         occ_stride=(8, 8)):  # pas du déplacement (H,W)
    x = x.requires_grad_(True)

    occ = Occlusion(model)
    ig  = IntegratedGradients(model)
    ixg = InputXGradient(model)

    # Occlusion a BESOIN de sliding_window_shapes
    attr_occ = occ.attribute(
        x,
        target=target,
        sliding_window_shapes=(3, occ_patch[0], occ_patch[1]),
        strides=(3, occ_stride[0], occ_stride[1]),
        baselines=0
    )  # (1,3,224,224)

    attr_ig  = ig.attribute(x, target=target, n_steps=50)      # (1,3,224,224)
    attr_ixg = ixg.attribute(x, target=target)                 # (1,3,224,224)

    def to_map(a):
        a = a.detach().cpu()[0]            # (3,224,224)
        a = a.abs().sum(dim=0).numpy()     # (224,224)
        return normalize_attr(a)

    return {
        "Occlusion": to_map(attr_occ),
        "IntegratedGradients": to_map(attr_ig),
        "InputXGradient": to_map(attr_ixg),
    }
```

```python
tests_images_urls = []
tests_images_urls += cat_images_urls[:2]
tests_images_urls += person_images_urls[:3]

for image_url in tests_images_urls:

    # 1) Load image
    image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB")
    image_np = np.array(image)

    # 2) Preprocess image
    input_tensor = preprocess(image)
    x = input_tensor.unsqueeze(0).to(DEVICE)

    # 3) Prediction (model returns logits)
    model.eval()
    logits = model(x)
    probabilities = F.softmax(logits, dim=1)
    predicted_class_index = probabilities[0].argmax(0).item()

    predicted_probability = probabilities[0, predicted_class_index].item()

    if imagenet_labels is not None:
        prediction_name = imagenet_labels[predicted_class_index]
    else:
        prediction_name = f"class_index={predicted_class_index}"

    print("\nImage:", image_url)
    print("predicted_class_index:", predicted_class_index)
    print("Predicted class:", prediction_name, "| probability:", round(predicted_probability, 3))

    # 4) Display original
    plt.figure(figsize=(7, 7))
    plt.imshow(image_np)
    plt.title(f'Predicted class: {prediction_name}, probability: {round(predicted_probability, 3)}')
    plt.axis('off')
    plt.show()

    # 5) Captum attributions (3 méthodes)
    attrs = compute_attributions(model, x, target=predicted_class_index)

    for method_name, attr_map in attrs.items():
        show_attributions(image, attr_map, title=f"{method_name} (target={prediction_name})")
```

#### Les méthodes Occlusion, Integrated Gradients et Input×Gradient servent à expliquer la prédiction en produisant un heatmap des pixels pour une classe cible .



#### En comparant les heatmaps, on observe que les méthodes ne mettent pas toujours en avant exactement les mêmes zones : elles capturent différentes notions d’importance. Une explication est jugée meilleure si les zones mises en évidence correspondent visuellement à l’objet pertinent et si la suppression des pixels importants fait diminuer la probabilité de la classe (test de faithfulness).


## (b) Implement the faithfulness test, and plot its results for those 3 methods for the sample images chosen in part (a)

```python
def deletion_curve(model, x, target, attr_map, steps=20):
    """
    Faithfulness: on supprime progressivement les pixels les + importants (selon attr_map)
    et on observe la baisse de proba de la classe cible.
    x: (1,3,224,224) normalisé
    attr_map: (224,224) importance normalisée [0,1]
    """
    model.eval()
    x0 = x.detach().clone()

    H, W = attr_map.shape
    flat = attr_map.reshape(-1)
    order = np.argsort(flat)[::-1]  # plus importants d'abord

    baseline = torch.zeros_like(x0)  # image "noire" dans l'espace normalisé

    fracs = np.linspace(0, 1, steps + 1)
    scores = []

    for frac in fracs:
        k = int(frac * H * W)
        mask = np.ones(H*W, dtype=np.float32)
        if k > 0:
            mask[order[:k]] = 0.0
        mask = torch.tensor(mask.reshape(H, W), device=x0.device).unsqueeze(0).unsqueeze(0)  # (1,1,H,W)
        mask3 = mask.repeat(1, 3, 1, 1)  # (1,3,H,W)

        x_occ = x0 * mask3 + baseline * (1 - mask3)

        with torch.no_grad():
            logits = model(x_occ)
            prob = F.softmax(logits, dim=1)[0, target].item()
        scores.append(prob)

    return fracs, np.array(scores)

def auc(fracs, scores):
    return float(np.trapz(scores, fracs))

def plot_curves(curves, title="Deletion curves"):
    plt.figure(figsize=(7,5))
    for name, (fracs, scores) in curves.items():
        plt.plot(fracs, scores, label=name)
    plt.xlabel("Fraction de pixels supprimés (les + importants)")
    plt.ylabel("Probabilité classe cible")
    plt.title(title)
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
for image_url in tests_images_urls:
    # Load image (style prof)
    image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB")
    image_np = np.array(image)

    # Preprocess
    input_tensor = preprocess(image)
    x = input_tensor.unsqueeze(0).to(DEVICE)

    # Prediction
    model.eval()
    logits = model(x)
    probs = F.softmax(logits, dim=1)
    target = probs[0].argmax(0).item()
    p = probs[0, target].item()

    target_name = imagenet_labels[target] if imagenet_labels else f"class_index={target}"
    print("\nImage:", image_url)
    print("Target (classe prédite):", target_name, "| prob=", round(p, 3))

    # Attributions (Occlusion/IG/IXG)
    attrs = compute_attributions(model, x, target=target)

    # Faithfulness curves + AUC
    curves = {}
    aucs = {}
    for method_name, attr_map in attrs.items():
        fracs, scores = deletion_curve(model, x, target, attr_map, steps=20)
        curves[method_name] = (fracs, scores)
        aucs[method_name] = auc(fracs, scores)

    plot_curves(curves, title=f"Model | Deletion curves | target={target_name}")

    print("AUC (plus petit = chute plus rapide = plus faithful):")
    for k,v in sorted(aucs.items(), key=lambda t: t[1]):
        print(f" - {k}: {v:.4f}")
```

#### Nous avons évalué la “faithfulness” des explications en supprimant progressivement les pixels les plus importants (selon chaque méthode) et en mesurant la baisse de la probabilité de la classe cible. Une AUC plus faible signifie que la probabilité chute plus rapidement  donc que la méthode identifie mieux les régions réellement utilisées par le modèle.

#### Sur nos exemples, Integrated Gradients et Input×Gradient obtiennent le plus souvent les AUC les plus faibles, donc des explications plus fidèles dans ce test. Occlusion est parfois compétitive (ex. sunglasses avec AUC très faible), mais elle peut aussi être moins performante (ex. ambulance où son AUC est nettement plus élevée). Enfin, les résultats varient selon l’image et la classe cible : il n’existe pas une méthode toujours meilleure, ce qui montre que la qualité d’explication dépend du contenu de l’image et du comportement du modèle.

## (d) Compare the results for both cases

```python
def evaluate_model_auc(model_eval, model_name, urls, steps=20):
    rows = []
    for image_url in urls:
        image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB")
        x = preprocess(image).unsqueeze(0).to(DEVICE)

        model_eval.eval()
        probs = F.softmax(model_eval(x), dim=1)
        target = probs[0].argmax(0).item()
        target_name = imagenet_labels[target] if imagenet_labels else f"class_index={target}"

        attrs = compute_attributions(model_eval, x, target=target)

        for method_name, attr_map in attrs.items():
            fracs, scores = deletion_curve(model_eval, x, target, attr_map, steps=steps)
            rows.append({
                "model": model_name,
                "image": image_url.split("/")[-1],
                "target": target_name,
                "method": method_name,
                "auc": auc(fracs, scores)
            })
    return rows

urls_eval = [cat_images_urls[1], person_images_urls[0]]

rows = []
rows += evaluate_model_auc(model,  "Model1", urls_eval, steps=20)
rows += evaluate_model_auc(model2, "Model2", urls_eval, steps=20)

# Affichage "table" simple (sans pandas)
print("\n=== Résultats AUC (plus petit = plus faithful en deletion) ===")
rows_sorted = sorted(rows, key=lambda r: (r["model"], r["image"], r["auc"]))
for r in rows_sorted:
    print(f'{r["model"]:6s} | {r["image"]:20s} | target={r["target"][:18]:18s} | {r["method"]:18s} | AUC={r["auc"]:.4f}')
```

#### Question (d) – Comparaison des méthodes et des modèles (AUC – deletion curves)

On compare la faithfulness des explications via l’AUC : plus l’AUC est faible, plus la probabilité de la classe cible chute rapidement quand on supprime les pixels “importants”, donc l’explication est plus fidèle dans ce test.

1) Variation selon l’image
- Image 000000000171.jpg :
  * Model1 (classe sunglasses) : Occlusion est la meilleure (AUC=0.0046), puis Integrated Gradients (0.0062), puis Input×Gradient (0.0120).
  * Model2 (classe suit) : Integrated Gradients est la meilleure (0.0098), puis Input×Gradient (0.0170), puis Occlusion (0.0205).
  → La méthode la plus fidèle dépend donc de l’image ET de la classe cible (et du modèle).

2) Variation selon le modèle (même image mais cible différente)
- Pour 000000000171.jpg, Model1 prédit “sunglasses” alors que Model2 prédit “suit”.
  Cela montre que deux modèles peuvent s’appuyer sur des indices différents et même ne pas reconnaître la même chose ; les explications et les scores de faithfulness changent donc.

3) Cas de 000000001600.jpg (toilet seat)
- Model1 : les AUC sont très élevées (~0.33), ce qui indique que supprimer les pixels “importants” ne fait pas chuter rapidement la probabilité → explications moins faithfulness (et/ou décision moins localisée sur une zone précise).
- Model2 : AUC plus faibles (IG=0.0678, IXG=0.0740) et Occlusion plus élevée (0.1620).
  → Pour MobileNetV2, les méthodes à base de gradients semblent plus fidèles sur cette image.

Conclusion :
La faithfulness dépend fortement du modèle et de l’image. Il n’y a pas une seule méthode toujours meilleure : sur certains cas Occlusion est très efficace (Model1 sur sunglasses), tandis que sur d’autres les méthodes gradient (Integrated Gradients, Input×Gradient) sont plus fidèles (Model2 sur toilet seat). Ce test confirme qu’il faut comparer plusieurs méthodes et plusieurs modèles.
