# Dice

```python
def dice(image_1, image_2):
  '''returns the dice score between two binary images'''
  intersection = ((image_1 == 1) & (image_2 == 1)).sum()
  image_1_non_zero = (image_1 != 0).sum()
  image_2_non_zero = (image_2 != 0).sum()
  score = 2 * intersection / (image_1_non_zero + image_2_non_zero)
  return score
```

# Setup and Sample Prediction for FCN

```python
# https://docs.pytorch.org/vision/main/models.html

from torchvision.models.segmentation import fcn_resnet50, FCN_ResNet50_Weights

# Model initialization

weights = FCN_ResNet50_Weights.DEFAULT
fcn_model = fcn_resnet50(weights=weights)
fcn_model.eval()

# Preprocessor

fcn_preprocess = weights.transforms()
```

```python
# Load image

image_url = cat_images_urls[0]

image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB")
image_np = np.array(image)

# Preprocess image

image = fcn_preprocess(image)

# Prediction (model returns a dictionary, relevant key is 'out')

prediction = fcn_model(image.unsqueeze(0))["out"]
print(prediction.shape)
predicted_mask = prediction[0].argmax(0)
predicted_mask = predicted_mask.detach().cpu().numpy()

fcn_cat_prediction = predicted_mask == cat_category

# Display

plt.figure(figsize = (10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_np)
plt.subplot(1, 2, 2)
plt.imshow(fcn_cat_prediction)
```

# Setup and Sample Prediction for CLIPSeg

```python
# https://huggingface.co/docs/transformers/en/model_doc/clipseg

import torch
from transformers import AutoProcessor, CLIPSegForImageSegmentation
from transformers.image_utils import load_image

# Model initialization
clip_seg_model = CLIPSegForImageSegmentation.from_pretrained("CIDAS/clipseg-rd64-refined")

# Preprocessor
clip_seg_processor = AutoProcessor.from_pretrained("CIDAS/clipseg-rd64-refined")
```

```python
# Load image

image_url = cat_images_urls[0]

image = load_image(image_url)
image_np = np.array(image)

# Preprocess image

text_prompts = ["a cat"]
image = clip_seg_processor(text=text_prompts, images=[image] * len(text_prompts), padding=True, return_tensors="pt")

# Prediction

with torch.inference_mode():
    output = clip_seg_model(**image)

prediction = output.logits
print(prediction.shape)

predicted_mask = prediction[0] > 0.5
clipseg_cat_prediction = predicted_mask.detach().cpu().numpy()

# Display

plt.figure(figsize = (10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image_np)
plt.subplot(1, 2, 2)
plt.imshow(clipseg_cat_prediction)
```

# Compare predictions for the remaining images both qualitatively (visualize) as well as quantitatively (compute similarity between predictions using dice)

```python
def loadandpredict(i):
  ###### FCN #####
  # Load image
  image_url = cat_images_urls[i]
  image = Image.open(requests.get(image_url, stream=True).raw).convert("RGB")
  image_np = np.array(image)

  # Preprocess image
  image_fcn = fcn_preprocess(image)

  # Prediction (model returns a dictionary, relevant key is 'out')
  prediction = fcn_model(image_fcn.unsqueeze(0))["out"]
  predicted_mask = prediction[0].argmax(0)
  predicted_mask = predicted_mask.detach().cpu().numpy()

  fcn_cat_prediction = predicted_mask == cat_category

  # RESIZE FCN prediction back to original size
  fcn_cat_prediction = cv2.resize(
      fcn_cat_prediction.astype(np.uint8),
      dsize=(image_np.shape[1], image_np.shape[0]),  # (width, height)
      interpolation=cv2.INTER_NEAREST
  )

  ##### Clip ######
  # Load image again for CLIP
  image_clip = load_image(image_url)

  # Preprocess image
  text_prompts = ["a cat"]
  image_processed = clip_seg_processor(text=text_prompts, images=[image_clip] * len(text_prompts), padding=True, return_tensors="pt")

  # Prediction
  with torch.inference_mode():
      output = clip_seg_model(**image_processed)

  prediction = output.logits
  predicted_mask = prediction[0] > 0.5
  clipseg_cat_prediction = predicted_mask.detach().cpu().numpy()

  clipseg_cat_prediction = cv2.resize(
      clipseg_cat_prediction.astype(np.uint8),
      dsize=(image_np.shape[1], image_np.shape[0]),  # (width, height)
      interpolation=cv2.INTER_NEAREST
  )


  ##### SAM (Segment Anything Model) ######
  # SAM doesn't need special preprocessing - it works directly on the image
  sam_image_rgb = cv2.cvtColor(image_np, cv2.COLOR_BGR2RGB)

  # Set the image for SAM predictor
  sam_predictor.set_image(sam_image_rgb)

  # Get image embeddings
  image_embedding = sam_predictor.get_image_embedding().cpu().numpy()

  # For automatic segmentation, we can use point prompts or bounding box
  # Here's a simple approach using the center point as prompt
  h, w = sam_image_rgb.shape[:2]

  # Use center point as prompt (assuming cat is roughly centered)
  input_point = np.array([[w//2, h//2]])
  input_label = np.array([1])  # 1 indicates foreground point

  # Predict segmentation mask
  masks, scores, logits = sam_predictor.predict(
      point_coords=input_point,
      point_labels=input_label,
      multimask_output=True,  # Get multiple masks
  )

  # Choose the best mask (highest score)
  best_mask_idx = np.argmax(scores)
  sam_cat_prediction = masks[best_mask_idx]

  return fcn_cat_prediction, clipseg_cat_prediction, sam_cat_prediction, image_np



def DiceAndDisplay(fcn_cat_prediction, clipseg_cat_prediction, sam_cat_prediction, image_np, GT_path):
    # Load ground truth mask
    GT_image = Image.open(GT_path).convert('L')  # Convert to grayscale
    GT_np = np.array(GT_image)

    # Convert GT to binary (assuming white=cat, black=background)
    GT_binary = (GT_np > 128).astype(np.uint8)

    # Convert all predictions to binary format if they're not already
    fcn_binary = fcn_cat_prediction.astype(np.uint8) if fcn_cat_prediction.dtype == bool else (fcn_cat_prediction > 0).astype(np.uint8)
    clip_binary = clipseg_cat_prediction.astype(np.uint8) if clipseg_cat_prediction.dtype == bool else (clipseg_cat_prediction > 0).astype(np.uint8)
    sam_binary = sam_cat_prediction.astype(np.uint8) if sam_cat_prediction.dtype == bool else (sam_cat_prediction > 0).astype(np.uint8)

    # Calculate dice scores
    fcn_dice = dice(fcn_binary, GT_binary)
    clip_dice = dice(clip_binary, GT_binary)
    sam_dice = dice(sam_binary, GT_binary)

    # Create 2x2 grid for display
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))

    # Original image
    axes[0, 0].imshow(image_np)
    axes[0, 0].set_title('Original Image')
    axes[0, 0].axis('off')

    # FCN prediction
    axes[0, 1].imshow(fcn_binary, cmap='gray')
    axes[0, 1].set_title(f'FCN Prediction\nDice Score: {fcn_dice:.3f}')
    axes[0, 1].axis('off')

    # CLIPSeg prediction
    axes[1, 0].imshow(clip_binary, cmap='gray')
    axes[1, 0].set_title(f'CLIPSeg Prediction\nDice Score: {clip_dice:.3f}')
    axes[1, 0].axis('off')

    # SAM prediction
    axes[1, 1].imshow(sam_binary, cmap='gray')
    axes[1, 1].set_title(f'SAM Prediction\nDice Score: {sam_dice:.3f}')
    axes[1, 1].axis('off')

    plt.tight_layout()
    plt.show()

    # Print scores for reference
    print(f"Dice Scores Summary:")
    print(f"  FCN: {fcn_dice:.3f}")
    print(f"  CLIPSeg: {clip_dice:.3f}")
    print(f"  SAM: {sam_dice:.3f}")

    return fcn_dice, clip_dice, sam_dice
```
